```python
import matplotlib.pyplot as plt
import numpy as np
import os
```

```python
path = "/home/clod/Desktop/THESIS_2/FULL_ATOM/RUNS/hybrid_displacement/small_models_no_ff/reconstruction_data" 

path_csv = os.path.join(path, "total_energy.csv")
total_energy_list = np.loadtxt(path_csv, delimiter=',', skiprows=1)


plt.figure(figsize=(10, 6))
plt.title("Total physics loss")
plt.xlabel("Epochs")
plt.ylabel("Total energy (kcal/mol)")
plt.yscale('log')
plt.plot(total_energy_list[:, 0], total_energy_list[:, 1], label='Total energy', color='blue')
plt.grid(alpha=0.5)
plt.legend()
plt.savefig(os.path.join(path, "total_energy_plot.png"), dpi=300)

plt.show()
```
